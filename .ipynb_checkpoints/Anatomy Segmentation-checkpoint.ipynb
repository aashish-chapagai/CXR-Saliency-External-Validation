```python
import os
import re
import sys
import glob
import warnings
import cv2
import torch
import numpy as np
import pandas as pd
import scipy.sparse as sp
import matplotlib.pyplot as plt
import pydicom
try:
    from pydicom.pixels import apply_modality_lut, apply_voi_lut
except ImportError:
    from pydicom.pixel_data_handlers.util import apply_modality_lut, apply_voi_lut    

os.makedirs('models', exist_ok=True)
os.makedirs('figures', exist_ok=True)
```

```python
HYBRIDGNET_DIR = os.path.join('external', 'hybridgnet')
CHEXLOCALIZE_DIR = os.path.join('external', 'chexlocalize')
CHEXMASK_DIR = os.path.join('external', 'chexmask')
VINDR_DIR = os.path.join('external', 'vindr')
NIH_IMAGE_DIRS = ['images', '/mnt/c/Users/AASHISH/Downloads/Sentic Team/Notebooks/images']
NIH_CHECK_IMAGES = 300
CHUNK_ROWS = 1000
ORGANS = ('Left Lung', 'Right Lung', 'Heart')
IMAGE_EXTENSIONS = ('.jpg', '.jpeg', '.png')
DICOM_EXTENSIONS = ('.dicom', '.dcm')
SEED = 42
CHEXPERT_ID = re.compile(r'(patient\d+)[/_](study\d+)[/_](view\d+_(?:frontal|lateral))', re.IGNORECASE)
```

```python
torch.set_num_threads(os.cpu_count())
sys.path.insert(0, HYBRIDGNET_DIR)

with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    from models.HybridGNet2IGSC import Hybrid
    from utils.utils import scipy_to_torch_sparse, genMatrixesLungsHeart
```

```python
def load_hybridgnet(weights_path):
    A, AD, D, U = genMatrixesLungsHeart()
    config = {
        'n_nodes': [A.shape[0]] * 3 + [AD.shape[0]] * 3,
        'latents': 64,
        'inputsize': 1024,
        'filters': [2, 32, 32, 32, 16, 16, 16],
        'skip_features': 32,
    }
    adjacency = [scipy_to_torch_sparse(sp.csc_matrix(m).tocoo()) for m in (A, A, A, AD, AD, AD)]
    downsample = [scipy_to_torch_sparse(sp.csc_matrix(D).tocoo())]
    upsample = [scipy_to_torch_sparse(sp.csc_matrix(U).tocoo())]
    network = Hybrid(config, downsample, upsample, adjacency)
    network.load_state_dict(torch.load(weights_path, map_location='cpu', weights_only=True))
    return network.eval()

with warnings.catch_warnings():
    warnings.simplefilter('ignore')
    hybridgnet = load_hybridgnet(os.path.join(HYBRIDGNET_DIR, 'weights', 'weights.pt'))
print(f"HybridGNet loaded ({sum(p.numel() for p in hybridgnet.parameters()):,} parameters), running on {torch.get_num_threads()} CPU threads.")
```

```python
def pad_to_square(image):
    height, width = image.shape
    if height > width:
        pad = height - width
        return np.pad(image, ((0, 0), (pad // 2, pad // 2 + pad % 2))), 0, pad
    pad = width - height
    return np.pad(image, ((pad // 2, pad // 2 + pad % 2), (0, 0))), pad, 0

def read_dicom(path):
    dataset = pydicom.dcmread(path)
    pixels = apply_modality_lut(dataset.pixel_array, dataset).round().astype(np.int32)
    pixels = apply_voi_lut(pixels, dataset).astype(np.float32)
    if str(dataset.get('PhotometricInterpretation', '')).upper() == 'MONOCHROME1':
        pixels = pixels.max() - pixels
    pixels -= pixels.min()
    return np.uint8(np.round(255 * pixels / max(float(pixels.max()), 1e-6)))

def read_radiograph(path):
    if os.path.splitext(path)[1].lower() in DICOM_EXTENSIONS:
        return read_dicom(path)
    return cv2.imread(path, cv2.IMREAD_GRAYSCALE)

def segment(path):
    image = read_radiograph(path) / 255.0
    padded, pad_h, pad_w = pad_to_square(image)
    size = padded.shape[0]
    resized = padded if size == 1024 else cv2.resize(padded, (1024, 1024), interpolation=cv2.INTER_CUBIC)
    with torch.no_grad(), warnings.catch_warnings():
        warnings.simplefilter('ignore')
        landmarks = hybridgnet(torch.from_numpy(resized).float()[None, None])[0].numpy().reshape(-1, 2) * size
    landmarks[:, 0] -= pad_w // 2
    landmarks[:, 1] -= pad_h // 2
    landmarks = landmarks.astype(np.int32)
    masks = {}
    for name, part in (('Right Lung', landmarks[:44]), ('Left Lung', landmarks[44:94]), ('Heart', landmarks[94:])):
        canvas = np.zeros(image.shape, dtype=np.uint8)
        masks[name] = cv2.drawContours(canvas, [part.reshape(-1, 1, 2)], -1, 1, -1).astype(bool)
    return masks
```

```python
def encode_rle(mask):
    pixels = np.concatenate([[0], mask.ravel().astype(np.int8), [0]])
    runs = np.flatnonzero(pixels[1:] != pixels[:-1]) + 1
    runs[1::2] -= runs[::2]
    return ' '.join(map(str, runs))

def decode_rle(rle, height, width):
    counts = np.zeros(height * width + 1, dtype=np.int32)
    if isinstance(rle, str) and rle.strip():
        runs = np.asarray(rle.split(), dtype=np.int64)
        np.add.at(counts, runs[0::2] - 1, 1)
        np.add.at(counts, runs[0::2] - 1 + runs[1::2], -1)
    return (np.cumsum(counts[:-1]) > 0).reshape(height, width)

def plausible(masks):
    left, right, heart = masks['Left Lung'], masks['Right Lung'], masks['Heart']
    if not (left.any() and right.any() and heart.any()):
        return 0.0
    left_y, left_x = np.nonzero(left)
    right_y, right_x = np.nonzero(right)
    side_by_side = abs(left_x.mean() - right_x.mean()) > abs(left_y.mean() - right_y.mean())
    return float(side_by_side and 0.1 <= (left | right).mean() <= 0.7)

def dice(first, second):
    total = first.sum() + second.sum()
    return 2.0 * np.logical_and(first, second).sum() / total if total else np.nan
```

```python
def chexpert_key(path):
    match = CHEXPERT_ID.search(str(path).replace('\\', '/'))
    return '_'.join(part.lower() for part in match.groups()) if match else None

def image_id(path):
    return os.path.splitext(os.path.basename(str(path)))[0]

def segment_all(images, name):
    rows = []
    for i, record in enumerate(images.to_dict('records')):
        masks = segment(record['path'])
        height, width = masks['Heart'].shape
        rows.append({
            'Path': record['key'],
            'Quality': plausible(masks),
            'Left Lung': encode_rle(masks['Left Lung']),
            'Right Lung': encode_rle(masks['Right Lung']),
            'Heart': encode_rle(masks['Heart']),
            'Height': height,
            'Width': width,
        })
        if i % 100 == 0:
            print(f"  Segmented {i} of {len(images)} {name} images", flush=True)
    table = pd.DataFrame(rows)
    output = f'models/hybridgnet_masks_{name.lower()}.csv'
    table.to_csv(output, index=False)
    print(f"Saved HybridGNet masks for {len(table)} {name} images to {output}.")
    print(f"Anatomically plausible: {int(table['Quality'].sum())} of {len(table)}.")
    return table
```

```python
records = {}
for path in sorted(glob.glob(os.path.join(CHEXLOCALIZE_DIR, '**', '*'), recursive=True)):
    key = chexpert_key(path) if os.path.splitext(path)[1].lower() in IMAGE_EXTENSIONS else None
    if key is not None and key.endswith('frontal') and key not in records:
        parts = path.replace('\\', '/').split('/')
        split = 'val' if 'val' in parts else 'test' if 'test' in parts else 'unknown'
        records[key] = {'key': key, 'path': path, 'split': split}
chexpert = pd.DataFrame(list(records.values()))
print(f"Found {len(chexpert)} frontal CheXpert images.")
print(f"Validation split: {int((chexpert['split'] == 'val').sum())} images.")
print(f"Test split: {int((chexpert['split'] == 'test').sum())} images.")
segmented = {'CheXpert': segment_all(chexpert, 'CheXpert')}
```

```python
label_files = glob.glob(os.path.join(VINDR_DIR, '**', 'image_labels_test.csv'), recursive=True)
if label_files:
    labels = pd.read_csv(label_files[0])
    id_column = {column.strip().lower(): column for column in labels.columns}.get('image_id', labels.columns[0])
    wanted = set(labels[id_column].astype(str))
    dicoms = {}
    for path in sorted(glob.glob(os.path.join(VINDR_DIR, '**', '*'), recursive=True)):
        if os.path.splitext(path)[1].lower() in DICOM_EXTENSIONS and image_id(path) in wanted:
            dicoms.setdefault(image_id(path), path)
    keys = sorted(dicoms)
    vindr = pd.DataFrame({'key': keys, 'path': [dicoms[key] for key in keys]})
    print(f"Found {len(vindr)} of {len(wanted)} labelled VinDr-CXR test images.")
    segmented['VinDr'] = segment_all(vindr, 'VinDr')
else:
    vindr = None
    print(f"Label file not found in {VINDR_DIR}. Skipping the VinDr segmentation.")
```

```python
def compare_with_chexmask(csv_path, ours, normalise, default_size=None):
    header = pd.read_csv(csv_path, nrows=0).columns.tolist()
    lookup = {name.strip().lower(): name for name in header}
    expected = ORGANS + ('Height', 'Width')
    columns = [header[0]] + [lookup[c.lower()] for c in expected if c.lower() in lookup]
    our_keys = set(ours)
    scores = []
    for chunk in pd.read_csv(csv_path, usecols=columns, chunksize=CHUNK_ROWS):
        keys = chunk[header[0]].astype(str).map(normalise)
        matched = keys.isin(our_keys).to_numpy()      
        for key, record in zip(keys[matched], chunk[matched].to_dict('records')):
            mine = ours[key]
            height = int(record[lookup['height']]) if 'height' in lookup else default_size
            width = int(record[lookup['width']]) if 'width' in lookup else default_size
            if (height, width) != (mine['Height'], mine['Width']):
                continue
            theirs = {organ: decode_rle(record[lookup[organ.lower()]], height, width) for organ in ORGANS}
            mine_masks = {organ: decode_rle(mine[organ], height, width) for organ in ORGANS}
            mine_lungs = mine_masks['Left Lung'] | mine_masks['Right Lung']
            their_lungs = theirs['Left Lung'] | theirs['Right Lung']
            scores.append({
                'key': key,
                'lungs': dice(mine_lungs, their_lungs),
                'heart': dice(mine_masks['Heart'], theirs['Heart']),
            })
    return pd.DataFrame(scores, columns=['key', 'lungs', 'heart'])

def report(name, scores):
    if len(scores) == 0:
        print(f"{name}: no overlapping images to compare.")
        return
    print(f"{name}: {len(scores)} images compared with CheXmask.")
    print(f"  Lungs Dice mean {scores['lungs'].mean():.3f}, 5th percentile {scores['lungs'].quantile(0.05):.3f}")
    print(f"  Heart Dice mean {scores['heart'].mean():.3f}, 5th percentile {scores['heart'].quantile(0.05):.3f}")

for name, csv_name, normalise in (('CheXpert', 'CheXpert.csv', chexpert_key), ('VinDr', 'VinDr-CXR.csv', image_id)):
    csv_path = os.path.join(CHEXMASK_DIR, csv_name)
    if name not in segmented:
        continue
    if not os.path.exists(csv_path):
        print(f"{csv_name} not found in {CHEXMASK_DIR}. Skipping the {name} agreement check.")
        continue
    scores = compare_with_chexmask(csv_path, segmented[name].set_index('Path').to_dict('index'), normalise)
    scores.to_csv(f'models/hybridgnet_agreement_{name.lower()}.csv', index=False)
    report(name, scores)
```

```python
nih_metadata = pd.read_csv('models/metadata_test.csv')
sample_size = min(NIH_CHECK_IMAGES, len(nih_metadata))
chosen = set(nih_metadata['Image Index'].sample(n=sample_size, random_state=SEED).astype(str))
nih_paths = {}
for directory in NIH_IMAGE_DIRS:
    if os.path.isdir(directory):
        for root, _, files in os.walk(directory):
            for name in files:
                if name in chosen:
                    nih_paths.setdefault(name, os.path.join(root, name))
if nih_paths:
    nih_ours = {}
    for name, path in nih_paths.items():
        masks = segment(path)
        height, width = masks['Heart'].shape
        nih_ours[name] = {
            'Left Lung': encode_rle(masks['Left Lung']),
            'Right Lung': encode_rle(masks['Right Lung']),
            'Heart': encode_rle(masks['Heart']),
            'Height': height,
            'Width': width,
        }
    nih_csv = os.path.join(CHEXMASK_DIR, 'ChestX-Ray8.csv')
    nih_scores = compare_with_chexmask(nih_csv, nih_ours, os.path.basename, default_size=1024)
    nih_scores.to_csv('models/hybridgnet_agreement_nih.csv', index=False)
    report('NIH test sample', nih_scores)
else:
    print("NIH original images not found in NIH_IMAGE_DIRS. Skipping the NIH agreement check.")
```

```python
panels = [('CheXpert', chexpert[chexpert['split'] == 'test'].head(3))]
if vindr is not None and len(vindr):
    panels.append(('VinDr', vindr.head(3)))
fig, axes = plt.subplots(len(panels), 3, figsize=(13.5, 4.8 * len(panels)), squeeze=False)
for axis_row, (name, images) in zip(axes, panels):
    table = segmented[name].set_index('Path')
    for ax, record in zip(axis_row, images.to_dict('records')):
        masks = table.loc[record['key']]
        ax.imshow(read_radiograph(record['path']), cmap='gray')
        for organ, colour in (('Left Lung', 'lime'), ('Right Lung', 'lime'), ('Heart', 'magenta')):
            mask = decode_rle(masks[organ], int(masks['Height']), int(masks['Width'])).astype(np.uint8)
            contours, _ = cv2.findContours(mask, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_NONE)
            for contour in contours:
                closed = np.vstack([contour[:, 0, :], contour[:1, 0, :]])
                ax.plot(closed[:, 0], closed[:, 1], color=colour, linewidth=1)
        ax.set_title(f"{name}: {record['key']} (plausible: {bool(masks['Quality'])})", fontsize=8)
    for ax in axis_row:
        ax.axis('off')
plt.suptitle('HybridGNet segmentation of external test images: lungs (green), heart (magenta)')
plt.tight_layout()
plt.savefig('figures/hybridgnet_check.png', dpi=300, bbox_inches='tight')
plt.show()
```
